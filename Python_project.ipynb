```python
#sale event avg performance
sale_summary = df.groupby("Sale_Event").agg({

    "Discount_Pct":"mean",

    "Current_Price_INR":"mean",

    "Reviews_Count":"sum"

}).round(2)

sale_summary
```

```python
#price segments
def price_segment(price):

    if price < 10000:
        return "Budget"

    elif price < 40000:
        return "Mid Range"

    elif price < 60000:
        return "Premium"

    else:
        return "Luxury"


df["Price_Segment"] = df["Current_Price_INR"].apply(price_segment)

df["Price_Segment"].value_counts()
```

```python
#KPIs
print("Total Products :", len(df))

print("Total Categories :", df["Product_Category"].nunique())

print("Total Platforms :", df["Platform"].nunique())

print("Average Launch Price :",
      round(df["Launch_Price_INR"].mean(),2))

print("Average Current Price :",
      round(df["Current_Price_INR"].mean(),2))

print("Average Discount :",
      round(df["Discount_Pct"].mean(),2),"%")

print("Average Rating :",
      round(df["Rating"].mean(),2))

print("Total Reviews :",df["Reviews_Count"].sum())
```
